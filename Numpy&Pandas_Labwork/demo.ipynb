```python
import numpy as np
arr3=[30,40,50]
arr4=np.array([[1,2,3],
               [4,5,6]])

print("Addition:")
print(arr3+arr4)

print("Subtraction:")
print(arr4 - arr3)

print("Subtraction:")
print(arr4 * arr3)

print("Division:")
print(arr4 / arr3)

print("Floor Division:")
print(arr4 // arr3)

print("Power:")
print(arr4 ** arr3)
```
