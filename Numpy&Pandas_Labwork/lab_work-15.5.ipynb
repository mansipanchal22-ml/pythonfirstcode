2. Handling Missing Values, Duplicates, and Data Types

Task 5: Load a DataFrame with missing values and fill them with appropriate values

```python
import numpy as np
import pandas as pd

data = {
    'Name': ['Aarav', 'Diya', 'Rohan', 'Priya', 'Kavya'],
    'Age': [25, np.nan, 28, np.nan, 22],
    'Salary': [50000, 60000, np.nan, 45000, 52000],
    'City': ['Ahmedabad', 'Surat', np.nan, 'Ahmedabad', np.nan],
}

df = pd.DataFrame(data)

print("--data--")
print(df)
print("values:")
print(df.isnull().sum())


df['Age'] = df['Age'].fillna(df['Age'].mean())

df['Salary'] = df['Salary'].fillna(df['Salary'].median())

df['City'] = df['City'].fillna('Unknown')

print("\n--data2--")
print(df)
```

Task 6: Drop rows where any value is missing

```python
df = pd.DataFrame(data)

df_cleaned = df.dropna()
print(df_cleaned)
```

Task 7: Identify and remove duplicate rows

```python
import pandas as pd

data = {
    'Name': ['Aarav', 'Diya', 'Aarav', 'Rohan'],
    'City': ['Ahmedabad', 'Surat', 'Ahmedabad', 'Vadodara'],
}

df = pd.DataFrame(data)
print("Duplicates rows count:", df.duplicated().sum())

df_cleaned = df.drop_duplicates()
print("Dataframe after removing duplicates:")
print(df_cleaned)
```

Task 8: Convert a column from string to numeric (handle errors gracefully)

```python
import pandas as pd

data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'Price': ['55000', '800', 'Not Available', '12000'],
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print("Original Data Type of Price:", df['Price'].dtype)

df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

print("\n-------------------------------------------")
print("DataFrame after conversion:")
print(df)
print("\nNew Data Type of Price:", df['Price'].dtype)
```

Task 9: Check and change data types of columns in a DataFrame


```python
import pandas as pd
data = {
    'order_id': [1001, 1002, 1003],
    'product_code': ['A1', 'B2', 'C3'],
    'quantity': ['5', '12', '8'],        # numbers stored as text
    'unit_price': [99.50, 150.00, 45.25],
    'delivered': [1, 0, 1]                # numeric flags (1/0)
}
df = pd.DataFrame(data)
print("--Initial Column Types--")
print(df.dtypes)

type_mapping = {
    'order_id': str,       
    'quantity': int,       
    'unit_price': float,
    'delivered': bool
}

df = df.astype(type_mapping)
print("--updated column types--")
print(df.dtypes)
print("--Dataframe--")
print(df)
```

3. Handling Dates and Times using Pandas

Task 10: Load a DataFrame containing a 'date' column with string format, and convert it to datetime

```python
import pandas as pd

data = {
    'order_id': [101, 102, 103, 104],
    'order_date': ['2024-01-10', '2024-02-15', '2024-03-20', '2024-04-25'],
    'delivery_date': ['2024-01-15', '2024-02-22', '2024-03-28', '2024-05-02'],
}

df = pd.DataFrame(data)
```

```python
#task-10
df['order_date'] = pd.to_datetime(df['order_date'])
df['delivery_date'] = pd.to_datetime(df['delivery_date'])

print(data)
print("new data:")
print(df.dtypes)
```

Task 11: Extract year, month, and day from the datetime column

```python
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['day'] = df['order_date'].dt.day

print("--Extracted Year, Month, Day--")
print(df[['order_id', 'order_date', 'year', 'month', 'day']])
```

Task 12: Filter the DataFrame for rows within a specific date range



```python
filtered_df = df[
    (df['order_date'] >= '2024-02-01') & (df['order_date'] <= '2024-03-31')
]

print("--Filtered Rows (Feb 1 to Mar 31) --")
print(filtered_df[['order_id', 'order_date']])
```

Task 13: Create a new column with the difference in days between two datetime columns


```python
df['delivery_days'] = (df['delivery_date'] - df['order_date']).dt.days

print("\n--Delivery Days Difference --")
print(df[['order_id', 'order_date', 'delivery_date', 'delivery_days']])
```
